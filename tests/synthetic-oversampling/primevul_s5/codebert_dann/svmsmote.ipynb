```python
"""
Cross-Project Vulnerability Detection on PrimeVul
--------------------------------------------------
Leave-One-CWE-Out (LOCO)
Train on ALL CWEs EXCEPT CWE-787 -> Test on CWE-787 (pooled, cross-project)
SVMSMOTE Oversampling, With Class Weighting, WITH Domain-Adversarial Neural Network (DANN)
"""

import os
import json
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.autograd import Function
from torch.utils.data import DataLoader, TensorDataset
from tqdm.notebook import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score,
    roc_auc_score, f1_score, confusion_matrix
)
from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import SVMSMOTE
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# =========================================================
# Config
# =========================================================
SEED         = 42
HELD_OUT_CWE = "CWE-787"
COMBINED_FILE = "../../../../embedding/primevul/codebert/primevul_embedded.jsonl"
RAW_FILE      = "../../../../dataset/primevul/primevul_raw_combined.jsonl"
EMB_KEY      = "emb"
OUTPUT_DIR   = f"results/svmsmote_with_classweight_dann/{HELD_OUT_CWE}"

# DANN-specific
LAMBDA_MAX   = 1.0   # max weight for domain loss (GRL strength ceiling)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.backends.mps.is_available():
    DEVICE = "mps"
    torch.mps.manual_seed(SEED)
elif torch.cuda.is_available():
    DEVICE = "cuda"
    torch.cuda.manual_seed_all(SEED)
else:
    DEVICE = "cpu"


# =========================================================
# CWE Lookup
# =========================================================
def load_cwe_lookup(raw_path):
    cwe_map = {}
    with open(raw_path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            rec = json.loads(line)
            cwe_list = rec.get("cwe", [])
            cwe_val = cwe_list[0] if cwe_list else None
            cwe_map[rec["idx"]] = cwe_val
    return cwe_map


# =========================================================
# Data Loading
# =========================================================
def load_jsonl_with_cwe(path, cwe_map):
    records = []
    with open(path, "r") as f:
        for line in f:
            line = line.strip()
            if line:
                records.append(json.loads(line))

    X        = np.array([r[EMB_KEY]   for r in records], dtype=np.float32)
    y        = np.array([r["target"]  for r in records], dtype=np.int32)
    projects = np.array([r["project"] for r in records], dtype=object)
    cwe      = np.array([cwe_map.get(r["idx"], None) for r in records], dtype=object)
    return X, y, projects, cwe


# =========================================================
# Gradient Reversal Layer
# =========================================================
class GradientReversalFunction(Function):
    @staticmethod
    def forward(ctx, x, lambda_):
        ctx.lambda_ = lambda_
        return x.view_as(x)

    @staticmethod
    def backward(ctx, grad_output):
        return -ctx.lambda_ * grad_output, None


class GradientReversalLayer(nn.Module):
    def __init__(self, lambda_=1.0):
        super().__init__()
        self.lambda_ = lambda_

    def forward(self, x):
        return GradientReversalFunction.apply(x, self.lambda_)

    def set_lambda(self, lambda_):
        self.lambda_ = lambda_


# =========================================================
# DANN Architecture: FeatureExtractor + ClassifierHead + DomainDiscriminator
# =========================================================
class FeatureExtractor(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
        )

    def forward(self, x):
        return self.network(x)  # shared embedding, dim=64


class ClassifierHead(nn.Module):
    def __init__(self, feature_dim=64):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(feature_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, features):
        return self.network(features)


class DomainDiscriminator(nn.Module):
    def __init__(self, feature_dim=64):
        super().__init__()
        self.grl = GradientReversalLayer(lambda_=1.0)
        self.network = nn.Sequential(
            nn.Linear(feature_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )

    def forward(self, features):
        reversed_features = self.grl(features)
        return self.network(reversed_features)


# =========================================================
# DANN Training Loop
# =========================================================
def train_dann(feature_extractor, classifier_head, domain_discriminator,
                x_source, y_source, x_target,
                x_val, y_val,
                class_weights, epochs, desc):
    """
    x_source/y_source : SVMSMOTE-oversampled labeled training pool (source domain)
    x_target          : UNLABELED held-out CWE test set (used only for domain adaptation)
    """
    pos_weight = torch.tensor(class_weights[1], dtype=torch.float32).to(DEVICE)
    neg_weight = torch.tensor(class_weights[0], dtype=torch.float32).to(DEVICE)

    params = (list(feature_extractor.parameters()) +
              list(classifier_head.parameters()) +
              list(domain_discriminator.parameters()))
    optimizer = torch.optim.Adam(params, lr=1e-5)

    label_criterion  = nn.BCELoss(reduction='none')
    domain_criterion = nn.BCELoss()

    source_dataset = TensorDataset(
        torch.tensor(x_source, dtype=torch.float32),
        torch.tensor(y_source, dtype=torch.float32)
    )
    target_dataset = TensorDataset(
        torch.tensor(x_target, dtype=torch.float32)
    )

    batch_size = 64
    source_loader = DataLoader(source_dataset, batch_size=batch_size, shuffle=True)
    target_loader = DataLoader(target_dataset, batch_size=batch_size, shuffle=True)

    n_batches = max(len(source_loader), len(target_loader))

    for epoch in tqdm(range(epochs), desc=desc, leave=False):
        feature_extractor.train()
        classifier_head.train()
        domain_discriminator.train()

        p = epoch / max(epochs - 1, 1)
        lambda_ = (2.0 / (1.0 + np.exp(-10 * p)) - 1.0) * LAMBDA_MAX
        domain_discriminator.grl.set_lambda(lambda_)

        source_iter = iter(source_loader)
        target_iter = iter(target_loader)

        for _ in range(n_batches):
            try:
                x_s, y_s = next(source_iter)
            except StopIteration:
                source_iter = iter(source_loader)
                x_s, y_s = next(source_iter)
            try:
                (x_t,) = next(target_iter)
            except StopIteration:
                target_iter = iter(target_loader)
                (x_t,) = next(target_iter)

            x_s, y_s = x_s.to(DEVICE), y_s.to(DEVICE)
            x_t = x_t.to(DEVICE)

            optimizer.zero_grad()

            # --- Label prediction loss (source only) ---
            features_s = feature_extractor(x_s)
            label_pred = classifier_head(features_s).squeeze()
            weights    = torch.where(y_s == 1, pos_weight, neg_weight)
            label_loss = (label_criterion(label_pred, y_s) * weights).mean()

            # --- Domain discrimination loss (source + target) ---
            features_t = feature_extractor(x_t)
            domain_pred_s = domain_discriminator(features_s).squeeze()
            domain_pred_t = domain_discriminator(features_t).squeeze()

            domain_labels_s = torch.zeros(domain_pred_s.shape[0]).to(DEVICE)  # source = 0
            domain_labels_t = torch.ones(domain_pred_t.shape[0]).to(DEVICE)   # target = 1

            domain_loss = (domain_criterion(domain_pred_s, domain_labels_s) +
                            domain_criterion(domain_pred_t, domain_labels_t))

            total_loss = label_loss + domain_loss
            total_loss.backward()
            optimizer.step()

        feature_extractor.eval()
        classifier_head.eval()
        with torch.no_grad():
            val_features = feature_extractor(x_val)
            val_preds    = classifier_head(val_features).squeeze()
            weights      = torch.where(y_val == 1, pos_weight, neg_weight)
            val_loss     = (label_criterion(val_preds, y_val) * weights).mean().item()

        print(f"      {desc} | Epoch {epoch+1}/{epochs} - "
              f"label_loss: {label_loss.item():.4f} - domain_loss: {domain_loss.item():.4f} - "
              f"val_loss: {val_loss:.4f} - lambda: {lambda_:.3f}")

    return feature_extractor, classifier_head, domain_discriminator


# =========================================================
# XGBoost (trained on domain-invariant FEATURES, not raw embeddings)
# =========================================================
def train_base_model_xgb(x_train, y_train, class_weights, sample_weights=None):
    scale_pos_weight = class_weights[1] / class_weights[0]
    model = XGBClassifier(
        n_estimators=100,
        max_depth=3,
        eval_metric='logloss',
        scale_pos_weight=scale_pos_weight,
        random_state=SEED
    )
    model.fit(x_train, y_train, sample_weight=sample_weights)
    return model


# =========================================================
# Confusion Matrix Plot (unchanged)
# =========================================================
def plot_confusion_matrix(tn, fp, fn, tp, save_path):
    cm = np.array([[tn, fp], [fn, tp]])
    fig, ax = plt.subplots(figsize=(6, 5))
    img = ax.imshow(cm, interpolation='nearest')
    ax.set_title(f"Confusion Matrix ({HELD_OUT_CWE} Test Set) - SVMSMOTE + DANN")
    plt.colorbar(img, ax=ax)
    tick_marks = np.arange(2)
    ax.set_xticks(tick_marks)
    ax.set_xticklabels(["Non-Vulnerable", "Vulnerable"])
    ax.set_yticks(tick_marks)
    ax.set_yticklabels(["Non-Vulnerable", "Vulnerable"])
    thresh = cm.max() / 2
    for i in range(2):
        for j in range(2):
            color = "black" if cm[i, j] > thresh else "white"
            ax.text(j, i, cm[i, j], ha="center", va="center", color=color)
    ax.set_ylabel("Actual Label")
    ax.set_xlabel("Predicted Label")
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    print(f"Confusion matrix saved as {save_path}")
    plt.show()


# =========================================================
# Main
# =========================================================
def main():
    print(f"\n=== PrimeVul | LOCO + SVMSMOTE + DANN: Train on ALL except {HELD_OUT_CWE} -> Test on {HELD_OUT_CWE} ===")
    print(f"    Device: {DEVICE}")

    print("\n[1/7] Building CWE lookup from raw file...")
    cwe_map = load_cwe_lookup(RAW_FILE)

    print("\n[2/7] Loading combined embeddings with CWE join...")
    X_all, y_all, proj_all, cwe_all = load_jsonl_with_cwe(COMBINED_FILE, cwe_map)

    n_missing = int(np.sum((y_all == 1) & (cwe_all == None)))
    if n_missing > 0:
        print(f"      WARNING: {n_missing} vulnerable rows had no CWE match via idx join.")

    print(f"\n[3/7] Splitting by CWE (held-out: {HELD_OUT_CWE})...")
    vuln_mask       = (y_all == 1)
    test_vuln_mask  = vuln_mask & (cwe_all == HELD_OUT_CWE)
    train_vuln_mask = vuln_mask & (cwe_all != HELD_OUT_CWE) & (cwe_all != None)

    if test_vuln_mask.sum() == 0:
        raise ValueError(f"No vulnerable instances found for {HELD_OUT_CWE}. Check CWE join.")

    benign_idx = np.where(y_all == 0)[0]
    benign_train_idx, benign_test_idx = train_test_split(
        benign_idx, test_size=0.2, random_state=SEED
    )

    train_idx = np.concatenate([np.where(train_vuln_mask)[0], benign_train_idx])
    test_idx  = np.concatenate([np.where(test_vuln_mask)[0],  benign_test_idx])

    train_emb = X_all[train_idx].astype(np.float32)
    train_lbl = y_all[train_idx].astype(np.int32)
    test_emb  = X_all[test_idx].astype(np.float32)
    test_lbl  = y_all[test_idx].astype(np.int32)  # only used for FINAL evaluation, never during training

    print(f"      Training pool (source, pre-SVMSMOTE) : {train_emb.shape}  "
          f"vuln={train_lbl.sum()}  benign={int((train_lbl == 0).sum())}")
    print(f"      Test set / target ({HELD_OUT_CWE})           : {test_emb.shape}  "
          f"vuln={test_lbl.sum()}  benign={int((test_lbl == 0).sum())}")
    print(f"      Training imbalance ratio             : {train_lbl.sum() / len(train_lbl):.4f}")

    print("\n[4/7] Normalizing embeddings...")
    scaler    = StandardScaler()
    train_emb = scaler.fit_transform(train_emb).astype(np.float32)
    test_emb  = scaler.transform(test_emb).astype(np.float32)

    print("\n[5/7] Applying SVMSMOTE to source training pool...")
    svmsmote = SVMSMOTE(random_state=SEED)
    train_emb, train_lbl = svmsmote.fit_resample(train_emb, train_lbl)
    train_emb = train_emb.astype(np.float32)
    train_lbl = train_lbl.astype(np.int32)
    print(f"      Training pool (post-SVMSMOTE)        : {train_emb.shape}  "
          f"vuln={train_lbl.sum()}  benign={int((train_lbl == 0).sum())}")

    print("\n[6/7] Computing class weights...")
    classes           = np.unique(train_lbl)
    class_weights_arr = compute_class_weight(class_weight='balanced', classes=classes, y=train_lbl)
    class_weight_dict = {int(c): float(w) for c, w in zip(classes, class_weights_arr)}
    print(f"      Class weights: {class_weight_dict}")

    print("\n[7/7] Running DANN + ensemble iterations...")
    num_iterations       = 3
    ensemble_predictions = np.zeros(len(test_lbl))

    for i in tqdm(range(num_iterations), desc="Ensemble", unit="iter"):
        print(f"\n      [Iteration {i+1}/{num_iterations}] Training DANN...")

        x_tr, x_val, y_tr, y_val = train_test_split(
            train_emb, train_lbl, test_size=0.2, random_state=SEED
        )
        x_val_t = torch.tensor(x_val, dtype=torch.float32).to(DEVICE)
        y_val_t = torch.tensor(y_val, dtype=torch.float32).to(DEVICE)

        feature_extractor     = FeatureExtractor(train_emb.shape[1]).to(DEVICE)
        classifier_head       = ClassifierHead(feature_dim=64).to(DEVICE)
        domain_discriminator  = DomainDiscriminator(feature_dim=64).to(DEVICE)

        feature_extractor, classifier_head, domain_discriminator = train_dann(
            feature_extractor, classifier_head, domain_discriminator,
            x_source=x_tr, y_source=y_tr,
            x_target=test_emb,          # unlabeled target domain, used only for domain loss
            x_val=x_val_t, y_val=y_val_t,
            class_weights=class_weight_dict,
            epochs=50,
            desc=f"Iter {i+1} - DANN"
        )

        feature_extractor.eval()
        with torch.no_grad():
            train_features = feature_extractor(
                torch.tensor(train_emb, dtype=torch.float32).to(DEVICE)
            ).cpu().numpy()
            test_features = feature_extractor(
                torch.tensor(test_emb, dtype=torch.float32).to(DEVICE)
            ).cpu().numpy()

            y_train_pred = classifier_head(
                torch.tensor(train_features, dtype=torch.float32).to(DEVICE)
            ).squeeze().cpu().numpy()

        print(f"      [Iteration {i+1}/{num_iterations}] Training XGBoost on DANN features...")
        sample_weights = np.where(train_lbl == 1, y_train_pred, 1 - y_train_pred)
        model_xgb      = train_base_model_xgb(
            train_features, train_lbl, class_weight_dict, sample_weights
        )
        model_xgb_pred        = model_xgb.predict_proba(test_features)[:, 1]
        ensemble_predictions += model_xgb_pred
        print(f"      [Iteration {i+1}/{num_iterations}] Done")

    print("\n[Metrics] Computing evaluation metrics...")
    ensemble_avg = ensemble_predictions / num_iterations
    y_pred_final = (ensemble_avg > 0.5).astype(int)

    accuracy  = accuracy_score(test_lbl,  y_pred_final)
    recall    = recall_score(test_lbl,    y_pred_final, zero_division=0)
    precision = precision_score(test_lbl, y_pred_final, zero_division=0)
    auc       = roc_auc_score(test_lbl,   ensemble_avg)
    f1        = f1_score(test_lbl,        y_pred_final, zero_division=0)

    tn, fp, fn, tp = confusion_matrix(test_lbl, y_pred_final).ravel()
    g_mean = np.sqrt((tp / (tp + fn + 1e-9)) * (tn / (tn + fp + 1e-9)))
    pf     = fp / (fp + tn + 1e-9)

    print(f"\n=== Evaluation Results ({HELD_OUT_CWE} Test Set) - SVMSMOTE + DANN ===")
    print(f"Accuracy  : {accuracy:.3f}")
    print(f"Precision : {precision:.3f}")
    print(f"Recall    : {recall:.3f}")
    print(f"F1-score  : {f1:.3f}")
    print(f"AUC       : {auc:.3f}")
    print(f"G-mean    : {g_mean:.3f}")
    print(f"PF value  : {pf:.3f}")
    print("\nConfusion Matrix:")
    print(f"  TN: {tn}  FP: {fp}")
    print(f"  FN: {fn}  TP: {tp}")

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    results = {
        "experiment"       : "svmsmote_with_classweight_dann",
        "scenario"         : "LOCO",
        "held_out_cwe"     : HELD_OUT_CWE,
        "n_train_samples"  : int(len(train_lbl)),
        "n_test_samples"   : int(len(test_lbl)),
        "n_vuln_train"     : int(train_lbl.sum()),
        "n_vuln_test"      : int(test_lbl.sum()),
        "accuracy"         : round(float(accuracy),  4),
        "precision"        : round(float(precision), 4),
        "recall"           : round(float(recall),    4),
        "f1"               : round(float(f1),        4),
        "auc"              : round(float(auc),        4),
        "g_mean"           : round(float(g_mean),    4),
        "pf"               : round(float(pf),        4),
        "confusion_matrix" : {
            "tn": int(tn), "fp": int(fp),
            "fn": int(fn), "tp": int(tp)
        }
    }

    json_path = os.path.join(OUTPUT_DIR, f"svmsmote_with_classweight_dann_{HELD_OUT_CWE}.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults saved to {json_path}")

    cm_path = os.path.join(OUTPUT_DIR, f"confusion_matrix_svmsmote_with_classweight_dann_{HELD_OUT_CWE}.png")
    plot_confusion_matrix(tn, fp, fn, tp, cm_path)


if __name__ == "__main__":
    main()
```
