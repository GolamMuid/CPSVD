```python
"""
Cross-Project Vulnerability Detection on ReVeal  (F1-oriented focal loss)
--------------------------------------------------------------------------
Train on Debian -> Test on Chrome
Focal Loss (algorithmic approach), no synthetic oversampling.
"""

import os
import json
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score,
    roc_auc_score, f1_score, confusion_matrix
)
from sklearn.utils.class_weight import compute_class_weight
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# =========================================================
# Config
# =========================================================
SEED        = 42
SOURCE_NAME = "Debian"
TARGET_NAME = "Chrome"

DATASET_DIR = "../../../../embedding/reveal/codebert"
OUTPUT_DIR  = "results/focal_loss/reveal_f1"

FOCAL_GAMMA = 2.0
ALPHA_GRID  = [0.25, 0.5, "derived"] 

NUM_ITERATIONS = 3
VAL_FRACTION   = 0.2  
THRESHOLD_GRID = np.linspace(0.05, 0.95, 91)

# XGBoost weighting switches 
USE_CONFIDENCE_WEIGHTS = True   
XGB_BASE_SCORE         = 0.0   

NN_EPOCHS_PHASE1 = 50
NN_EPOCHS_PHASE2 = 30
LOG_EVERY        = 10        

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.backends.mps.is_available():
    DEVICE = "mps"
    torch.mps.manual_seed(SEED)
elif torch.cuda.is_available():
    DEVICE = "cuda"
    torch.cuda.manual_seed_all(SEED)
else:
    DEVICE = "cpu"


# =========================================================
# Data
# =========================================================
def load_domain(name):
    prefix = name.lower()
    X = np.load(os.path.join(DATASET_DIR, f"{prefix}_embeddings.npy")).astype(np.float32)
    y = np.load(os.path.join(DATASET_DIR, f"{prefix}_labels.npy")).astype(np.int32)
    return X, y


# =========================================================
# Neural network + focal loss
# =========================================================
class VulnerabilityClassifier(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 256), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(64, 32), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(32, 16), nn.ReLU(),
            nn.Linear(16, 1), nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x)


class FocalLoss(nn.Module):
    def __init__(self, alpha=0.5, gamma=2.0, eps=1e-7):
        super().__init__()
        self.alpha, self.gamma, self.eps = alpha, gamma, eps

    def forward(self, predictions, targets):
        p = predictions.clamp(self.eps, 1 - self.eps)
        p_t = torch.where(targets == 1, p, 1 - p)
        alpha_t = torch.where(targets == 1,
                              torch.tensor(self.alpha, device=predictions.device),
                              torch.tensor(1 - self.alpha, device=predictions.device))
        return -alpha_t * (1 - p_t) ** self.gamma * torch.log(p_t)


def train_neural_network(model, dataloader, x_val, y_val, optimizer,
                         criterion, epochs, desc):
    for epoch in tqdm(range(epochs), desc=desc, leave=False):
        model.train()
        for x_b, y_b in dataloader:
            x_b, y_b = x_b.to(DEVICE), y_b.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(x_b).view(-1), y_b).mean()
            loss.backward()
            optimizer.step()

        if (epoch + 1) % LOG_EVERY == 0 or epoch == epochs - 1:
            model.eval()
            with torch.no_grad():
                v = criterion(model(x_val).view(-1), y_val).mean().item()
            print(f"      {desc} | Epoch {epoch+1}/{epochs} - val_loss: {v:.4f}")
    return model


def semi_supervised_transfer_learning(x_train, y_train, x_target,
                                      focal_criterion, iteration):
    x_tr, x_val, y_tr, y_val = train_test_split(
        x_train, y_train, test_size=0.2, random_state=SEED
    )
    x_tr_t   = torch.tensor(x_tr,  dtype=torch.float32)
    y_tr_t   = torch.tensor(y_tr,  dtype=torch.float32)
    x_val_t  = torch.tensor(x_val, dtype=torch.float32).to(DEVICE)
    y_val_t  = torch.tensor(y_val, dtype=torch.float32).to(DEVICE)
    x_tgt_t  = torch.tensor(x_target, dtype=torch.float32)

    loader = DataLoader(TensorDataset(x_tr_t, y_tr_t), batch_size=64, shuffle=True)
    model = VulnerabilityClassifier(x_train.shape[1]).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)

    model = train_neural_network(model, loader, x_val_t, y_val_t, optimizer,
                                 focal_criterion, NN_EPOCHS_PHASE1,
                                 f"Iter {iteration} - Phase 1")

    model.eval()
    with torch.no_grad():
        y_pseudo = (model(x_tgt_t.to(DEVICE)).view(-1).cpu().numpy() > 0.5).astype(int)

    x_aug = torch.tensor(np.concatenate((x_train, x_target)), dtype=torch.float32)
    y_aug = torch.tensor(np.concatenate((y_train, y_pseudo)), dtype=torch.float32)
    loader_aug = DataLoader(TensorDataset(x_aug, y_aug), batch_size=32, shuffle=True)

    model = train_neural_network(model, loader_aug, x_val_t, y_val_t, optimizer,
                                 focal_criterion, NN_EPOCHS_PHASE2,
                                 f"Iter {iteration} - Phase 2")
    return model


# =========================================================
# XGBoost with focal-loss objective
# =========================================================
def make_xgb_focal_loss_obj(alpha, gamma):
    eps = 1e-7

    def pointwise_focal(z, y):
        p = np.clip(1.0 / (1.0 + np.exp(-z)), eps, 1 - eps)
        p_t = np.where(y == 1, p, 1 - p)
        a_t = np.where(y == 1, alpha, 1 - alpha)
        return -a_t * (1 - p_t) ** gamma * np.log(p_t)

    def focal_obj(y_true, y_pred, sample_weight=None):
        y = np.asarray(y_true, dtype=np.float64)
        z = np.asarray(y_pred, dtype=np.float64)
        dz = 1e-4
        f_p, f_0, f_m = (pointwise_focal(z + dz, y), pointwise_focal(z, y),
                         pointwise_focal(z - dz, y))
        grad = (f_p - f_m) / (2 * dz)
        hess = np.clip((f_p - 2 * f_0 + f_m) / (dz ** 2), 1e-6, None)
        if sample_weight is not None:
            sw = np.asarray(sample_weight, dtype=np.float64)
            grad, hess = grad * sw, hess * sw
        return grad, hess

    return focal_obj


def train_base_model_xgb(x_train, y_train, alpha, gamma, sample_weights=None):
    model = XGBClassifier(
        n_estimators=100,
        max_depth=3,
        objective=make_xgb_focal_loss_obj(alpha, gamma),
        base_score=XGB_BASE_SCORE,
        eval_metric="logloss",
        random_state=SEED,
    )
    model.fit(x_train, y_train, sample_weight=sample_weights)
    return model


def predict_proba_focal(model, x):
    margin = model.predict(x, output_margin=True)
    return 1.0 / (1.0 + np.exp(-margin))


# =========================================================
# One full pipeline run for a given alpha
# =========================================================
def run_config(alpha, gamma, X_tr, y_tr, X_val, y_val, X_target):
    criterion = FocalLoss(alpha=alpha, gamma=gamma)
    val_sum  = np.zeros(len(y_val))
    test_sum = np.zeros(len(X_target))

    for i in range(NUM_ITERATIONS):
        print(f"\n      [alpha={alpha:.3f}] Iteration {i+1}/{NUM_ITERATIONS}: neural network...")
        model_nn = semi_supervised_transfer_learning(X_tr, y_tr, X_target, criterion, i + 1)

        model_nn.eval()
        with torch.no_grad():
            p_train = model_nn(torch.tensor(X_tr, dtype=torch.float32).to(DEVICE)
                               ).view(-1).cpu().numpy()

        if USE_CONFIDENCE_WEIGHTS:
            sample_weights = np.where(y_tr == 1, p_train, 1 - p_train)
        else:
            sample_weights = None

        print(f"      [alpha={alpha:.3f}] Iteration {i+1}/{NUM_ITERATIONS}: XGBoost...")
        model_xgb = train_base_model_xgb(X_tr, y_tr, alpha, gamma, sample_weights)
        val_sum  += predict_proba_focal(model_xgb, X_val)
        test_sum += predict_proba_focal(model_xgb, X_target)

    return val_sum / NUM_ITERATIONS, test_sum / NUM_ITERATIONS


# =========================================================
# Metrics / threshold selection
# =========================================================
def best_threshold_by_f1(y_true, prob):
    scores = [f1_score(y_true, (prob > t).astype(int), zero_division=0)
              for t in THRESHOLD_GRID]
    k = int(np.argmax(scores))
    return float(THRESHOLD_GRID[k]), float(scores[k])


def compute_metrics(y_true, prob, thr):
    pred = (prob > thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    return {
        "threshold": round(float(thr), 3),
        "accuracy":  round(float(accuracy_score(y_true, pred)), 4),
        "precision": round(float(precision_score(y_true, pred, zero_division=0)), 4),
        "recall":    round(float(recall_score(y_true, pred, zero_division=0)), 4),
        "f1":        round(float(f1_score(y_true, pred, zero_division=0)), 4),
        "auc":       round(float(roc_auc_score(y_true, prob)), 4),
        "g_mean":    round(float(np.sqrt((tp / (tp + fn + 1e-9)) * (tn / (tn + fp + 1e-9)))), 4),
        "pf":        round(float(fp / (fp + tn + 1e-9)), 4),
        "confusion_matrix": {"tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp)},
    }


def plot_confusion_matrix(cm_dict, title, save_path):
    cm = np.array([[cm_dict["tn"], cm_dict["fp"]], [cm_dict["fn"], cm_dict["tp"]]])
    fig, ax = plt.subplots(figsize=(6, 5))
    img = ax.imshow(cm, interpolation="nearest")
    ax.set_title(title)
    plt.colorbar(img, ax=ax)
    ticks = np.arange(2)
    ax.set_xticks(ticks); ax.set_xticklabels(["Non-Vulnerable", "Vulnerable"])
    ax.set_yticks(ticks); ax.set_yticklabels(["Non-Vulnerable", "Vulnerable"])
    thresh = cm.max() / 2
    for i in range(2):
        for j in range(2):
            ax.text(j, i, cm[i, j], ha="center", va="center",
                    color="black" if cm[i, j] > thresh else "white")
    ax.set_ylabel("Actual Label"); ax.set_xlabel("Predicted Label")
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    print(f"Confusion matrix saved as {save_path}")
    plt.show()


# =========================================================
# Main
# =========================================================
def main():
    print(f"\n=== ReVeal | {SOURCE_NAME} -> {TARGET_NAME} | focal loss, F1-oriented ===")
    print(f"    Device: {DEVICE}")

    print("\n[1/5] Loading embeddings...")
    src_emb, src_lbl = load_domain(SOURCE_NAME)
    tgt_emb, tgt_lbl = load_domain(TARGET_NAME)
    print(f"      {SOURCE_NAME}: {src_emb.shape}  vuln={int(src_lbl.sum())}")
    print(f"      {TARGET_NAME}: {tgt_emb.shape}  vuln={int(tgt_lbl.sum())}")

    print("\n[2/5] Normalising and creating source validation split...")
    scaler  = StandardScaler()
    src_emb = scaler.fit_transform(src_emb).astype(np.float32)
    tgt_emb = scaler.transform(tgt_emb).astype(np.float32)

    X_tr, X_val, y_tr, y_val = train_test_split(
        src_emb, src_lbl, test_size=VAL_FRACTION,
        stratify=src_lbl, random_state=SEED
    )
    print(f"      Source train: {X_tr.shape} (vuln={int(y_tr.sum())}) | "
          f"source val: {X_val.shape} (vuln={int(y_val.sum())})")

    cw = compute_class_weight("balanced", classes=np.unique(y_tr), y=y_tr)
    derived_alpha = float(cw[1] / (cw[1] + cw[0]))
    alphas = [derived_alpha if a == "derived" else float(a) for a in ALPHA_GRID]
    print(f"\n[3/5] Alpha candidates: {[round(a, 4) for a in alphas]} | gamma={FOCAL_GAMMA}")
    print(f"      Confidence weights on XGBoost: {USE_CONFIDENCE_WEIGHTS} | "
          f"base_score: {XGB_BASE_SCORE}")

    print("\n[4/5] Running configurations...")
    all_runs = []
    for alpha in alphas:
        val_prob, tgt_prob = run_config(alpha, FOCAL_GAMMA, X_tr, y_tr,
                                        X_val, y_val, tgt_emb)
        thr, val_f1 = best_threshold_by_f1(y_val, val_prob)
        run = {
            "alpha": round(alpha, 4),
            "gamma": FOCAL_GAMMA,
            "source_val_best_threshold": round(thr, 3),
            "source_val_f1_at_best_threshold": round(val_f1, 4),
            "target_at_0.5":       compute_metrics(tgt_lbl, tgt_prob, 0.5),
            "target_at_tuned_thr": compute_metrics(tgt_lbl, tgt_prob, thr),
            "_tgt_prob": tgt_prob,
        }
        all_runs.append(run)
        print(f"\n      alpha={alpha:.3f} | source-val F1={val_f1:.4f} @thr={thr:.2f} | "
              f"Chrome F1 @0.5={run['target_at_0.5']['f1']:.4f}, "
              f"@tuned={run['target_at_tuned_thr']['f1']:.4f}")

    best = max(all_runs, key=lambda r: r["source_val_f1_at_best_threshold"])
    print(f"\n[5/5] Selected alpha (by SOURCE validation F1): {best['alpha']}")

    for label, key in [("threshold 0.5", "target_at_0.5"),
                       ("source-tuned threshold", "target_at_tuned_thr")]:
        m = best[key]
        print(f"\n=== {TARGET_NAME} results, alpha={best['alpha']}, {label} ({m['threshold']}) ===")
        for k in ["accuracy", "precision", "recall", "f1", "auc", "g_mean", "pf"]:
            print(f"{k:<10}: {m[k]:.3f}")
        cm = m["confusion_matrix"]
        print(f"TN={cm['tn']} FP={cm['fp']} FN={cm['fn']} TP={cm['tp']}")

    os.makedirs(OUTPUT_DIR, exist_ok=True)
    tag = f"{SOURCE_NAME.lower()}_to_{TARGET_NAME.lower()}"

    out = {
        "experiment": "focal_loss_f1_oriented_no_oversampling",
        "dataset": "reveal",
        "train": SOURCE_NAME, "test": TARGET_NAME,
        "use_confidence_weights": USE_CONFIDENCE_WEIGHTS,
        "xgb_base_score": XGB_BASE_SCORE,
        "n_source_train": int(len(y_tr)), "n_source_val": int(len(y_val)),
        "n_target": int(len(tgt_lbl)),
        "selected_alpha": best["alpha"],
        "runs": [{k: v for k, v in r.items() if k != "_tgt_prob"} for r in all_runs],
    }
    json_path = os.path.join(OUTPUT_DIR, f"focal_loss_f1_{tag}.json")
    with open(json_path, "w") as f:
        json.dump(out, f, indent=2)
    print(f"\nResults saved to {json_path}")

    plot_confusion_matrix(
        best["target_at_tuned_thr"]["confusion_matrix"],
        f"Confusion Matrix ({TARGET_NAME}) - Focal Loss, alpha={best['alpha']}, "
        f"thr={best['target_at_tuned_thr']['threshold']}",
        os.path.join(OUTPUT_DIR, f"confusion_matrix_focal_f1_{tag}.png"),
    )


if __name__ == "__main__":
    main()
```
