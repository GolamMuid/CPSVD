```python
"""
Cross-Project Vulnerability Detection on PrimeVul
--------------------------------------------------
Train on ALL projects EXCEPT linux -> Test on linux
Tversky Loss (Algorithmic Approach), No Synthetic Oversampling
REVISED v4: fixes checkpoint selection gaming the loss function's
"predict everything positive" degenerate solution

Reference: Salehi, S. S. M., Erdogmus, D., & Gholipour, A. (2017).
"Tversky loss function for image segmentation using 3D fully convolutional
deep networks." MICCAI Workshop on Machine Learning in Medical Imaging.

WHAT v3 GOT WRONG:
v3 selected the "best" checkpoint by lowest validation Tversky loss. But
under the rate-normalized formulation with alpha=0.3, "predict every sample
positive" is itself a low-loss, stable fixed point:
    TP_rate = 1, FN_rate = 0, FP_rate = 1
    TI = 1 / (1 + alpha*1 + beta*0) = 1 / (1 + alpha)
    Loss = 1 - TI = alpha / (1 + alpha)   ~= 0.231 when alpha = 0.3
This is a mathematically legitimate low point of the loss surface, not a
training artifact -- so selecting checkpoints by loss alone will happily
lock onto this degenerate solution the moment training drifts near it, as
confirmed by the v3 result (TN=0, FP=52699, precision == test set's base
positive rate, G-mean == 0.000).

CHANGES IN THIS REVISION:

1. Checkpoint selection metric changed from validation Tversky loss ->
   validation G-mean (geometric mean of sensitivity and specificity, at a
   0.5 threshold on validation predictions). G-mean crashes to exactly 0.0
   the instant a model predicts a single class for everything, so it cannot
   be gamed by the same degenerate solution that fooled the raw loss value.
   The BEST checkpoint is now the one with the HIGHEST validation G-mean
   seen during training, not the lowest loss.

2. TVERSKY_ALPHA raised from 0.3 -> 0.45 (still beta > alpha, still biased
   toward recall, but the gap is narrowed so "ignore false positives
   entirely" is a less attractive fixed point for the loss surface itself
   to fall into). TVERSKY_BETA lowered correspondingly from 0.7 -> 0.55 to
   keep alpha+beta closer to a comparable overall scale to before.

3. Training loss (used for backprop) is UNCHANGED -- still the raw Tversky
   loss driving gradients each batch. Only the CHECKPOINT SELECTION
   criterion changes. This keeps the optimization objective consistent
   with "Tversky loss" as the technique under test, while using a
   degenerate-solution-proof metric purely to decide which epoch's weights
   to keep.

4. All v3 fixes retained: batch_size=512, rate-normalized loss, lr=1e-4,
   gradient clipping (max_norm=1.0), zero-init output bias.
"""

import os
import json
import copy
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from tqdm.notebook import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score,
    roc_auc_score, f1_score, confusion_matrix
)
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# =========================================================
# Config
# =========================================================
SEED        = 42
TEST_PROJ   = "linux"
COMBINED_FILE = "../../../../embedding/primevul/codebert/primevul_embedded.jsonl"
EMB_KEY     = "emb"
OUTPUT_DIR  = "results/tversky_loss/linux"

TVERSKY_ALPHA  = 0.45    # raised from 0.3 -- less exploitable by all-positive collapse
TVERSKY_BETA   = 0.55    # lowered from 0.7, still beta > alpha (recall-biased)
TVERSKY_SMOOTH = 1e-6

NN_LEARNING_RATE = 1e-4
BATCH_SIZE       = 512
GRAD_CLIP_NORM   = 1.0

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.backends.mps.is_available():
    DEVICE = "mps"
    torch.mps.manual_seed(SEED)
elif torch.cuda.is_available():
    DEVICE = "cuda"
    torch.cuda.manual_seed_all(SEED)
else:
    DEVICE = "cpu"


# =========================================================
# Data Loading (unchanged)
# =========================================================
def load_jsonl(path):
    records = []
    with open(path, "r") as f:
        for line in f:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    X        = np.array([r[EMB_KEY]   for r in records], dtype=np.float32)
    y        = np.array([r["target"]  for r in records], dtype=np.int32)
    projects = np.array([r["project"] for r in records], dtype=object)
    return X, y, projects


# =========================================================
# Neural Network (bias zero-init retained)
# =========================================================
class VulnerabilityClassifier(nn.Module):
    def __init__(self, input_dim):
        super(VulnerabilityClassifier, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )
        final_linear = self.network[-2]
        nn.init.zeros_(final_linear.bias)

    def forward(self, x):
        return self.network(x)


# =========================================================
# Rate-Normalized Tversky Loss (unchanged formula, new alpha/beta)
# =========================================================
class TverskyLoss(nn.Module):
    def __init__(self, alpha=0.45, beta=0.55, smooth=1e-6):
        super().__init__()
        self.alpha  = alpha
        self.beta   = beta
        self.smooth = smooth

    def forward(self, predictions, targets):
        p = predictions.view(-1)
        y = targets.view(-1)

        n_pos = y.sum().clamp(min=1.0)
        n_neg = (1 - y).sum().clamp(min=1.0)

        TP = (p * y).sum()
        FP = (p * (1 - y)).sum()
        FN = ((1 - p) * y).sum()

        TP_rate = TP / n_pos
        FP_rate = FP / n_neg
        FN_rate = FN / n_pos

        tversky_index = (TP_rate + self.smooth) / (
            TP_rate + self.alpha * FP_rate + self.beta * FN_rate + self.smooth
        )
        return 1.0 - tversky_index


# =========================================================
# G-mean helper (degenerate-solution-proof checkpoint metric)
# =========================================================
def compute_gmean(y_true, y_pred_binary):
    tp = int(np.sum((y_true == 1) & (y_pred_binary == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred_binary == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred_binary == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred_binary == 0)))
    sensitivity = tp / (tp + fn + 1e-9)
    specificity = tn / (tn + fp + 1e-9)
    return float(np.sqrt(sensitivity * specificity))


# =========================================================
# Training Loop: gradient clipping + BEST-BY-G-MEAN checkpointing
# =========================================================
def train_neural_network(model, dataloader, x_val, y_val,
                          optimizer, criterion, epochs, desc,
                          diagnostic_epochs=None, grad_clip_norm=1.0):
    diagnostic_epochs = diagnostic_epochs or set()
    y_val_np = y_val.cpu().numpy()

    best_gmean      = -1.0
    best_val_loss   = None
    best_state_dict = copy.deepcopy(model.state_dict())
    best_epoch      = 0

    for epoch in tqdm(range(epochs), desc=desc, leave=False):
        model.train()
        for x_batch, y_batch in dataloader:
            x_batch = x_batch.to(DEVICE)
            y_batch = y_batch.to(DEVICE)
            optimizer.zero_grad()
            predictions = model(x_batch).squeeze()
            loss        = criterion(predictions, y_batch)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=grad_clip_norm)
            optimizer.step()

        model.eval()
        with torch.no_grad():
            val_preds = model(x_val).squeeze()
            val_loss  = criterion(val_preds, y_val).item()
            val_preds_np = val_preds.cpu().numpy()
            val_gmean    = compute_gmean(y_val_np, (val_preds_np > 0.5).astype(int))

        print(f"      {desc} | Epoch {epoch+1}/{epochs} - val_loss: {val_loss:.4f} - val_gmean: {val_gmean:.4f}")

        if val_gmean > best_gmean:
            best_gmean      = val_gmean
            best_val_loss   = val_loss
            best_state_dict = copy.deepcopy(model.state_dict())
            best_epoch      = epoch + 1

        if (epoch + 1) in diagnostic_epochs:
            print(f"      [MID-TRAIN DIAGNOSTIC {desc} @ epoch {epoch+1}] "
                  f"val_loss: {val_loss:.4f}, val_gmean: {val_gmean:.4f}, "
                  f"val_pred min: {val_preds_np.min():.6f}, max: {val_preds_np.max():.6f}, "
                  f"mean: {val_preds_np.mean():.6f}  "
                  f"(best so far: epoch {best_epoch}, gmean {best_gmean:.4f})")
            if val_preds_np.max() < 1e-6 or val_preds_np.min() > 1 - 1e-6:
                print(f"      [WARNING] Output has collapsed to a degenerate "
                      f"constant by epoch {epoch+1}. G-mean-based checkpoint "
                      f"selection will avoid restoring this epoch since its "
                      f"G-mean is necessarily 0.")

    print(f"      {desc} | Restoring best checkpoint: epoch {best_epoch}, "
          f"val_gmean {best_gmean:.4f}, val_loss {best_val_loss:.4f} "
          f"(final epoch was {epochs}, val_loss {val_loss:.4f}, val_gmean {val_gmean:.4f})")
    model.load_state_dict(best_state_dict)
    return model


# =========================================================
# Semi-supervised Transfer Learning (Tversky Loss throughout)
# =========================================================
def semi_supervised_transfer_learning(x_train, y_train, x_test,
                                       tversky_criterion, iteration):
    x_tr, x_val, y_tr, y_val = train_test_split(
        x_train, y_train,
        test_size=0.2,
        random_state=SEED
    )

    x_tr_t   = torch.tensor(x_tr,   dtype=torch.float32)
    y_tr_t   = torch.tensor(y_tr,   dtype=torch.float32)
    x_val_t  = torch.tensor(x_val,  dtype=torch.float32).to(DEVICE)
    y_val_t  = torch.tensor(y_val,  dtype=torch.float32).to(DEVICE)
    x_test_t = torch.tensor(x_test, dtype=torch.float32)

    dataset    = TensorDataset(x_tr_t, y_tr_t)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

    model     = VulnerabilityClassifier(x_train.shape[1]).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=NN_LEARNING_RATE)

    # Phase 1: Train on source (all-except-linux)
    model = train_neural_network(
        model, dataloader, x_val_t, y_val_t,
        optimizer, tversky_criterion,
        epochs=50,
        desc=f"Iter {iteration} - Phase 1",
        diagnostic_epochs={10, 25},
        grad_clip_norm=GRAD_CLIP_NORM
    )

    # Pseudo-label the test set (linux)
    model.eval()
    with torch.no_grad():
        y_pred        = model(x_test_t.to(DEVICE)).squeeze().cpu().numpy()
        y_pred_binary = (y_pred > 0.5).astype(int)

    print(f"      [DIAGNOSTIC after Phase 1] y_pred -- min: {y_pred.min():.6f}, "
          f"max: {y_pred.max():.6f}, mean: {y_pred.mean():.6f}, "
          f"pseudo-labels flagged vulnerable: {y_pred_binary.sum()} / {len(y_pred_binary)}")

    if y_pred.max() - y_pred.min() < 1e-6:
        print("      [WARNING] Phase 1 best-checkpoint output is still a "
              "degenerate constant across the whole test set even after "
              "G-mean-based selection -- flag for further investigation "
              "before Phase 2.")

    # Phase 2: Fine-tune on source + pseudo-labelled linux
    x_train_aug    = np.concatenate((x_train, x_test))
    y_train_aug    = np.concatenate((y_train, y_pred_binary))
    x_aug_t        = torch.tensor(x_train_aug, dtype=torch.float32)
    y_aug_t        = torch.tensor(y_train_aug, dtype=torch.float32)
    x_aug_val_t    = torch.tensor(x_val, dtype=torch.float32).to(DEVICE)
    y_aug_val_t    = torch.tensor(y_val, dtype=torch.float32).to(DEVICE)
    dataset_aug    = TensorDataset(x_aug_t, y_aug_t)
    dataloader_aug = DataLoader(dataset_aug, batch_size=BATCH_SIZE, shuffle=True)

    model = train_neural_network(
        model, dataloader_aug, x_aug_val_t, y_aug_val_t,
        optimizer, tversky_criterion,
        epochs=30,
        desc=f"Iter {iteration} - Phase 2",
        diagnostic_epochs={10, 20},
        grad_clip_norm=GRAD_CLIP_NORM
    )

    model.eval()
    with torch.no_grad():
        y_pred_final = model(x_test_t.to(DEVICE)).squeeze().cpu().numpy()

    print(f"      [DIAGNOSTIC after Phase 2] y_pred_final -- min: {y_pred_final.min():.6f}, "
          f"max: {y_pred_final.max():.6f}, mean: {y_pred_final.mean():.6f}")

    return model, y_pred_final


# =========================================================
# Rate-normalized Tversky objective for XGBoost (unchanged formula,
# new alpha/beta passed in from config)
# =========================================================
def make_xgb_tversky_obj(alpha, beta, smooth=1e-6):
    def tversky_obj(y_true, y_pred, sample_weight=None):
        y = np.asarray(y_true, dtype=np.float64)
        z = np.asarray(y_pred, dtype=np.float64)
        p = 1.0 / (1.0 + np.exp(-z))

        n_pos = max(np.sum(y), 1.0)
        n_neg = max(np.sum(1 - y), 1.0)

        TP = np.sum(p * y)
        FP = np.sum(p * (1 - y))
        FN = np.sum((1 - p) * y)

        TP_rate = TP / n_pos
        FP_rate = FP / n_neg
        FN_rate = FN / n_pos

        denom = TP_rate + alpha * FP_rate + beta * FN_rate + smooth

        dTI_dTPrate = (denom - (TP_rate + smooth)) / (denom ** 2)
        dTI_dFPrate = -alpha * (TP_rate + smooth) / (denom ** 2)
        dTI_dFNrate = -beta  * (TP_rate + smooth) / (denom ** 2)

        dTI_dp = np.where(
            y == 1,
            dTI_dTPrate / n_pos - dTI_dFNrate / n_pos,
            dTI_dFPrate / n_neg
        )
        dloss_dp = -dTI_dp
        grad = dloss_dp * p * (1 - p)
        hess = p * (1 - p)
        hess = np.clip(hess, 1e-6, None)

        if sample_weight is not None:
            sw = np.asarray(sample_weight, dtype=np.float64)
            grad = grad * sw
            hess = hess * sw

        return grad, hess

    return tversky_obj


def train_base_model_xgb(x_train, y_train, alpha, beta, sample_weights=None):
    tversky_obj = make_xgb_tversky_obj(alpha, beta, smooth=TVERSKY_SMOOTH)
    model = XGBClassifier(
        n_estimators=100,
        max_depth=3,
        objective=tversky_obj,
        eval_metric='logloss',
        random_state=SEED
    )
    model.fit(x_train, y_train, sample_weight=sample_weights)
    return model


# =========================================================
# Confusion Matrix Plot
# =========================================================
def plot_confusion_matrix(tn, fp, fn, tp, save_path):
    cm = np.array([[tn, fp],
                   [fn, tp]])

    fig, ax = plt.subplots(figsize=(6, 5))
    img = ax.imshow(cm, interpolation='nearest')
    ax.set_title("Confusion Matrix (linux Test Set) - Tversky Loss (v4, G-mean checkpointed)")
    plt.colorbar(img, ax=ax)

    tick_marks = np.arange(2)
    ax.set_xticks(tick_marks)
    ax.set_xticklabels(["Non-Vulnerable", "Vulnerable"])
    ax.set_yticks(tick_marks)
    ax.set_yticklabels(["Non-Vulnerable", "Vulnerable"])

    thresh = cm.max() / 2
    for i in range(2):
        for j in range(2):
            color = "black" if cm[i, j] > thresh else "white"
            ax.text(j, i, cm[i, j], ha="center", va="center", color=color)

    ax.set_ylabel("Actual Label")
    ax.set_xlabel("Predicted Label")
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    print(f"Confusion matrix saved as {save_path}")
    plt.show()


# =========================================================
# Main
# =========================================================
def main():
    print(f"\n=== PrimeVul | Train on ALL except linux -> Test on linux (Tversky Loss, v4 G-mean checkpointed) ===")
    print(f"    Device: {DEVICE}")

    print("\n[1/5] Loading combined embeddings...")
    X_all, y_all, proj_all = load_jsonl(COMBINED_FILE)

    train_mask = proj_all != TEST_PROJ
    train_emb  = X_all[train_mask].astype(np.float32)
    train_lbl  = y_all[train_mask].astype(np.int32)

    test_mask = proj_all == TEST_PROJ
    test_emb  = X_all[test_mask].astype(np.float32)
    test_lbl  = y_all[test_mask].astype(np.int32)

    if train_emb.shape[0] == 0:
        raise ValueError("Training pool is empty. Check COMBINED_FILE path.")
    if test_emb.shape[0] == 0:
        raise ValueError(f"No samples found for '{TEST_PROJ}' in {COMBINED_FILE}.")

    print(f"      Training pool             : {train_emb.shape}  "
          f"vuln={train_lbl.sum()}  benign={int((train_lbl == 0).sum())}")
    print(f"      Projects in training pool : {len(set(proj_all[train_mask]))}")
    print(f"      Test set (linux)          : {test_emb.shape}  "
          f"vuln={test_lbl.sum()}  benign={int((test_lbl == 0).sum())}")
    pos_rate = train_lbl.sum() / len(train_lbl)
    print(f"      Training imbalance ratio  : {pos_rate:.4f}")

    print("\n[2/5] Normalizing embeddings...")
    scaler    = StandardScaler()
    train_emb = scaler.fit_transform(train_emb).astype(np.float32)
    test_emb  = scaler.transform(test_emb).astype(np.float32)

    print("\n[3/5] Setting up rate-normalized Tversky loss (v4: G-mean checkpointing + rebalanced alpha/beta)...")
    print(f"      alpha (FP rate weight) = {TVERSKY_ALPHA}  (raised from 0.3)")
    print(f"      beta  (FN rate weight) = {TVERSKY_BETA}  (lowered from 0.7)")
    print(f"      smooth                 = {TVERSKY_SMOOTH}")
    print(f"      NN learning rate       = {NN_LEARNING_RATE}")
    print(f"      Batch size             = {BATCH_SIZE}")
    print(f"      Gradient clip max_norm = {GRAD_CLIP_NORM}")
    print(f"      Checkpoint selection   = best validation G-mean (was: lowest val loss)")
    print(f"      All-positive-collapse loss value at these alpha/beta: "
          f"{TVERSKY_ALPHA / (1 + TVERSKY_ALPHA):.4f}  "
          f"(this fixed point can no longer be selected as 'best' since its G-mean is 0)")

    tversky_criterion = TverskyLoss(alpha=TVERSKY_ALPHA, beta=TVERSKY_BETA,
                                     smooth=TVERSKY_SMOOTH)

    print("\n[4/5] Running ensemble iterations...")
    num_iterations       = 3
    ensemble_predictions = np.zeros(len(test_lbl))

    for i in tqdm(range(num_iterations), desc="Ensemble", unit="iter"):
        print(f"\n      [Iteration {i+1}/{num_iterations}] Training neural network (Tversky Loss, G-mean checkpointed)...")
        model_nn, _ = semi_supervised_transfer_learning(
            train_emb, train_lbl,
            test_emb,
            tversky_criterion,
            iteration=i+1
        )

        print(f"      [Iteration {i+1}/{num_iterations}] Training XGBoost (Tversky objective)...")
        model_nn.eval()
        with torch.no_grad():
            x_tr_t       = torch.tensor(train_emb, dtype=torch.float32).to(DEVICE)
            y_train_pred = model_nn(x_tr_t).squeeze().cpu().numpy()

        sample_weights = np.where(train_lbl == 1, y_train_pred, 1 - y_train_pred)
        model_xgb      = train_base_model_xgb(
            train_emb, train_lbl, TVERSKY_ALPHA, TVERSKY_BETA, sample_weights
        )
        model_xgb_pred        = model_xgb.predict_proba(test_emb)[:, 1]
        ensemble_predictions += model_xgb_pred
        print(f"      [Iteration {i+1}/{num_iterations}] Done")

    print("\n[5/5] Computing metrics...")
    ensemble_avg = ensemble_predictions / num_iterations
    y_pred_final = (ensemble_avg > 0.5).astype(int)

    accuracy  = accuracy_score(test_lbl,  y_pred_final)
    recall    = recall_score(test_lbl,    y_pred_final, zero_division=0)
    precision = precision_score(test_lbl, y_pred_final, zero_division=0)
    auc       = roc_auc_score(test_lbl,   ensemble_avg)
    f1        = f1_score(test_lbl,        y_pred_final, zero_division=0)

    tn, fp, fn, tp = confusion_matrix(test_lbl, y_pred_final).ravel()
    g_mean = np.sqrt((tp / (tp + fn + 1e-9)) * (tn / (tn + fp + 1e-9)))
    pf     = fp / (fp + tn + 1e-9)

    print("\n=== Evaluation Results (linux Test Set) - Tversky Loss (v4) ===")
    print(f"Accuracy  : {accuracy:.3f}")
    print(f"Precision : {precision:.3f}")
    print(f"Recall    : {recall:.3f}")
    print(f"F1-score  : {f1:.3f}")
    print(f"AUC       : {auc:.3f}")
    print(f"G-mean    : {g_mean:.3f}")
    print(f"PF value  : {pf:.3f}")
    print("\nConfusion Matrix:")
    print(f"  TN: {tn}  FP: {fp}")
    print(f"  FN: {fn}  TP: {tp}")

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    results = {
        "experiment"       : "tversky_loss_no_oversampling_v4_gmean_checkpointed",
        "train"            : "all_except_linux",
        "test"             : TEST_PROJ,
        "tversky_alpha"    : TVERSKY_ALPHA,
        "tversky_beta"     : TVERSKY_BETA,
        "tversky_smooth"   : TVERSKY_SMOOTH,
        "nn_learning_rate" : NN_LEARNING_RATE,
        "batch_size"       : BATCH_SIZE,
        "grad_clip_norm"   : GRAD_CLIP_NORM,
        "checkpoint_metric": "validation_gmean",
        "n_train_projects" : int(len(set(proj_all[train_mask]))),
        "n_train_samples"  : int(len(train_lbl)),
        "n_test_samples"   : int(len(test_lbl)),
        "n_vuln_train"     : int(train_lbl.sum()),
        "n_vuln_test"      : int(test_lbl.sum()),
        "accuracy"         : round(float(accuracy),  4),
        "precision"        : round(float(precision), 4),
        "recall"           : round(float(recall),    4),
        "f1"               : round(float(f1),        4),
        "auc"              : round(float(auc),        4),
        "g_mean"           : round(float(g_mean),    4),
        "pf"               : round(float(pf),        4),
        "confusion_matrix" : {
            "tn": int(tn), "fp": int(fp),
            "fn": int(fn), "tp": int(tp)
        }
    }

    json_path = os.path.join(OUTPUT_DIR, "tversky_loss_no_oversampling_linux_v4.json")
    with open(json_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults saved to {json_path}")

    cm_path = os.path.join(OUTPUT_DIR, "confusion_matrix_tversky_loss_linux_v4.png")
    plot_confusion_matrix(tn, fp, fn, tp, cm_path)


if __name__ == "__main__":
    main()
```
